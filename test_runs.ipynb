```python
# print total number of distinct MONTH from the 'date' column, the format of dates is YYYY-MM-DD
df_enrol['date'] = pd.to_datetime(df_enrol['date'])
df_enrol['enrolment_month'] = df_enrol['date'].dt.to_period('M')
distinct_months = df_enrol['enrolment_month'].nunique()
print(f"Total number of distinct months in enrolment data: {distinct_months}")
```
